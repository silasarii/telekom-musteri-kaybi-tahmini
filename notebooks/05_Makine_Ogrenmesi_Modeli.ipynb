```python
# Gerekli kütüphaneler 

import pandas as pd

from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import roc_auc_score

import matplotlib.pyplot as plt
```

```python
# Modelleme için hazırlanmış veri setlerinin yüklenmesi
X_train = pd.read_csv("../data/X_train.csv")
X_test = pd.read_csv("../data/X_test.csv")

y_train = pd.read_csv("../data/y_train.csv").squeeze()
y_test = pd.read_csv("../data/y_test.csv").squeeze()

print("X_train boyutu:", X_train.shape)
print("X_test boyutu:", X_test.shape)
print("y_train boyutu:", y_train.shape)
print("y_test boyutu:", y_test.shape)
```

```python
# Lojistik Regresyon modelinin oluşturulması
lojistik_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

# Modelin eğitim verisiyle eğitilmesi
lojistik_model.fit(X_train, y_train)

print("Lojistik Regresyon modeli başarıyla eğitildi.")
```

```python
# Test verisi üzerinde tahmin
y_pred = lojistik_model.predict(X_test)

print("Tahminler başarıyla oluşturuldu.")
print("İlk 20 tahmin:")
print(y_pred[:20])
```

```python
# Confusion Matrix
cm = confusion_matrix(y_test, y_pred)

print("Confusion Matrix:")
print(cm)
```

```python
# Model performans metrikleri
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)

print("Lojistik Regresyon Modeli Performansı")
print("-------------------------------------")
print(f"Accuracy : {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall   : {recall:.4f}")
print(f"F1-score : {f1:.4f}")
```

```python
# KNN modelinin oluşturulması
knn_model = KNeighborsClassifier(n_neighbors=5)

# Modelin eğitilmesi
knn_model.fit(X_train, y_train)

print("KNN modeli başarıyla eğitildi.")
```

```python
# Test verisi üzerinde KNN tahminleri
y_pred_knn = knn_model.predict(X_test)

print("KNN tahminleri başarıyla oluşturuldu.")
print("İlk 20 tahmin:")
print(y_pred_knn[:20])
```

```python
# KNN Confusion Matrix
cm_knn = confusion_matrix(y_test, y_pred_knn)

print("KNN Confusion Matrix:")
print(cm_knn)
```

```python
# KNN model performans metrikleri
accuracy_knn = accuracy_score(y_test, y_pred_knn)
precision_knn = precision_score(y_test, y_pred_knn)
recall_knn = recall_score(y_test, y_pred_knn)
f1_knn = f1_score(y_test, y_pred_knn)

print("KNN Modeli Performansı")
print("----------------------")
print(f"Accuracy : {accuracy_knn:.4f}")
print(f"Precision: {precision_knn:.4f}")
print(f"Recall   : {recall_knn:.4f}")
print(f"F1-score : {f1_knn:.4f}")
```

```python
# Karar Ağacı modelinin oluşturulması
karar_agaci_model = DecisionTreeClassifier(
    random_state=42
)

# Modelin eğitilmesi
karar_agaci_model.fit(X_train, y_train)

print("Karar Ağacı modeli başarıyla eğitildi.")
```

```python
# Test verisi üzerinde Karar Ağacı tahminleri
y_pred_tree = karar_agaci_model.predict(X_test)

print("Karar Ağacı tahminleri başarıyla oluşturuldu.")
print("İlk 20 tahmin:")
print(y_pred_tree[:20])
```

```python
# Karar Ağacı Confusion Matrix
cm_tree = confusion_matrix(y_test, y_pred_tree)

print("Karar Ağacı Confusion Matrix:")
print(cm_tree)
```

```python
# Karar Ağacı model performans metrikleri
accuracy_tree = accuracy_score(y_test, y_pred_tree)
precision_tree = precision_score(y_test, y_pred_tree)
recall_tree = recall_score(y_test, y_pred_tree)
f1_tree = f1_score(y_test, y_pred_tree)

print("Karar Ağacı Modeli Performansı")
print("-----------------------------")
print(f"Accuracy : {accuracy_tree:.4f}")
print(f"Precision: {precision_tree:.4f}")
print(f"Recall   : {recall_tree:.4f}")
print(f"F1-score : {f1_tree:.4f}")
```

```python
# Random Forest modelinin oluşturulması
random_forest_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

# Modelin eğitilmesi
random_forest_model.fit(X_train, y_train)

print("Random Forest modeli başarıyla eğitildi.")
```

```python
# Test verisi üzerinde Random Forest tahminleri
y_pred_rf = random_forest_model.predict(X_test)

print("Random Forest tahminleri başarıyla oluşturuldu.")
print("İlk 20 tahmin:")
print(y_pred_rf[:20])
```

```python
# Random Forest Confusion Matrix
cm_rf = confusion_matrix(y_test, y_pred_rf)

print("Random Forest Confusion Matrix:")
print(cm_rf)
```

```python
# Random Forest model performans metrikleri
accuracy_rf = accuracy_score(y_test, y_pred_rf)
precision_rf = precision_score(y_test, y_pred_rf)
recall_rf = recall_score(y_test, y_pred_rf)
f1_rf = f1_score(y_test, y_pred_rf)

print("Random Forest Modeli Performansı")
print("--------------------------------")
print(f"Accuracy : {accuracy_rf:.4f}")
print(f"Precision: {precision_rf:.4f}")
print(f"Recall   : {recall_rf:.4f}")
print(f"F1-score : {f1_rf:.4f}")
```

```python
# Lojistik Regresyon için churn olasılıkları
y_prob = lojistik_model.predict_proba(X_test)[:, 1]

# ROC-AUC
roc_auc = roc_auc_score(y_test, y_prob)

print(f"Lojistik Regresyon ROC-AUC: {roc_auc:.4f}")
```

```python
# Lojistik Regresyon katsayılarının incelenmesi

katsayilar = pd.DataFrame({
    "Degisken": X_train.columns,
    "Katsayi": lojistik_model.coef_[0]
})

# Mutlak katsayı değerine göre sıralama
katsayilar["Mutlak_Katsayi"] = katsayilar["Katsayi"].abs()

katsayilar = katsayilar.sort_values(
    by="Mutlak_Katsayi",
    ascending=False
)

print("Lojistik Regresyon - En etkili değişkenler:")
print(katsayilar.head(15))
```

```python
# En etkili 15 değişken
grafik_verisi = katsayilar.head(15).sort_values("Katsayi")

plt.figure(figsize=(10, 7))

plt.barh(
    grafik_verisi["Degisken"],
    grafik_verisi["Katsayi"]
)

plt.axvline(0, linewidth=1)

plt.xlabel("Lojistik Regresyon Katsayısı")
plt.ylabel("Değişken")
plt.title("Lojistik Regresyon Modelinde En Etkili 15 Değişken")

plt.tight_layout()
plt.show()
```

```python
# Modellerin performans karşılaştırması

model_karsilastirma = pd.DataFrame({
    "Model": [
        "Lojistik Regresyon",
        "KNN",
        "Karar Ağacı",
        "Random Forest"
    ],
    "Accuracy": [
        accuracy,
        accuracy_knn,
        accuracy_tree,
        accuracy_rf
    ],
    "Precision": [
        precision,
        precision_knn,
        precision_tree,
        precision_rf
    ],
    "Recall": [
        recall,
        recall_knn,
        recall_tree,
        recall_rf
    ],
    "F1-score": [
        f1,
        f1_knn,
        f1_tree,
        f1_rf
    ]
})

print(model_karsilastirma.round(4))
```

```python
# Model performanslarının görselleştirilmesi

model_karsilastirma.set_index("Model")[
    ["Accuracy", "Precision", "Recall", "F1-score"]
].plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("Makine Öğrenmesi Modellerinin Performans Karşılaştırması")
plt.xlabel("Model")
plt.ylabel("Skor")
plt.xticks(rotation=0)
plt.ylim(0, 1)
plt.legend(title="Metrik")

plt.tight_layout()
plt.show()
```

```python
# Model karşılaştırma sonuçlarının kaydedilmesi

model_karsilastirma.to_csv(
    "../data/model_karsilastirma.csv",
    index=False
)

print("Model karşılaştırma sonuçları kaydedildi.")
```

```python
# Lojistik Regresyon tahminlerinin kaydedilmesi

final_tahminler = pd.DataFrame({
    "Gercek_Churn": y_test.values,
    "Tahmin_Churn": y_pred,
    "Churn_Olasiligi": y_prob
})

final_tahminler.to_csv(
    "../data/lojistik_regresyon_tahminleri.csv",
    index=False
)

print("Lojistik Regresyon tahminleri kaydedildi.")
```

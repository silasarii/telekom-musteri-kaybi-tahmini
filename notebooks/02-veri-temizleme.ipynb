```python
# Gerekli kütüphaneler
import numpy as np
import pandas as pd
```

```python
# Veri setinin okunması
veri = pd.read_csv("../data/WA_Fn-UseC_-Telco-Customer-Churn.csv")
```

```python
# Eksik değerleri kontrol etme
veri.isnull().sum()
```

```python
# Yinelenen satırları kontrol etme
veri.duplicated().sum()
```

```python
# TotalCharges sütunundaki boş string değerlerini kontrol etme
(veri["TotalCharges"] == "").sum()
```

```python
# TotalCharges sütunundaki sadece boşluk içeren değerleri kontrol etme
(veri["TotalCharges"].str.strip() == "").sum()
```

```python
# TotalCharges sütunundaki boşlukları NaN değerine dönüştürme
veri["TotalCharges"] = veri["TotalCharges"].replace(r'^\s*$', np.nan, regex=True)
```

```python
# Güncel eksik değer sayılarını kontrol etme
veri.isnull().sum()
```

```python
# TotalCharges değeri eksik olan kayıtları görüntüleme
veri[veri["TotalCharges"].isnull()]
```

```python
# TotalCharges değeri eksik olan kayıtları veri setinden kaldırma
veri = veri.dropna(subset=["TotalCharges"])
```

```python
# Veri setinin yeni boyutunu kontrol etme
veri.shape
```

```python
# TotalCharges sütununu sayısal veri tipine dönüştürme
veri["TotalCharges"] = pd.to_numeric(veri["TotalCharges"])
```

```python
# Veri tiplerini kontrol etme
veri.info()
```

```python
# Veri setinde eksik değer olup olmadığını kontrol etme
veri.isnull().sum()
```

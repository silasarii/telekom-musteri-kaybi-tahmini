```python
# Gerekli kütüphaneler 

import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
# Temizlenmiş veri setinin yüklenmesi
veri = pd.read_csv("../data/telco_temiz.csv")
```

```python
# Veri setinin ilk 5 satırını görüntüleme
veri.head()
```

```python
# Veri setinin boyutunu görüntüleme
veri.shape
```

```python
# Veri tiplerinin kontrolü
veri.dtypes
```

```python
# Müşteri kimlik numarasını veri setinden çıkarma

veri = veri.drop("customerID", axis=1)

# Güncel veri setinin boyutunu kontrol et

print("Güncel veri setinin boyutu:", veri.shape)
```

```python
# Hedef değişkenin sayısallaştırılması

veri["Churn"] = veri["Churn"].map({"No": 0, "Yes": 1})

# Kontrol

print(veri["Churn"].value_counts())
```

```python
# Kategorik değişkenlerin belirlenmesi

kategorik_degiskenler = veri.select_dtypes(include="object").columns

print("Kategorik değişkenler:")
print(kategorik_degiskenler)
```

```python
# Kategorik değişkenlerin One-Hot Encoding ile dönüştürülmesi

veri = pd.get_dummies(
    veri,
    columns=kategorik_degiskenler,
    drop_first=True,
    dtype=int
)

# Veri setinin boyutunu kontrol et

print("One-Hot Encoding sonrası veri setinin boyutu:", veri.shape)
```

```python
# Bağımsız değişkenler (X) ve hedef değişkenin (y) ayrılması

X = veri.drop("Churn", axis=1)
y = veri["Churn"]

# Boyutları kontrol et

print("X boyutu:", X.shape)
print("y boyutu:", y.shape)
```

```python
# Veriyi eğitim ve test kümelerine ayırma

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Boyutları kontrol et

print("X_train boyutu:", X_train.shape)
print("X_test boyutu:", X_test.shape)
print("y_train boyutu:", y_train.shape)
print("y_test boyutu:", y_test.shape)
```

```python
# Eğitim ve test kümelerinde Churn dağılımını kontrol et


print("Eğitim verisi Churn dağılımı:")
print(y_train.value_counts())
print(y_train.value_counts(normalize=True) * 100)

print("\nTest verisi Churn dağılımı:")
print(y_test.value_counts())
print(y_test.value_counts(normalize=True) * 100)
```

```python
# Ölçeklendirilecek sayısal değişkenler

sayisal_degiskenler = [
    "SeniorCitizen",
    "tenure",
    "MonthlyCharges",
    "TotalCharges"
]

# Ölçekleyiciyi oluştur

scaler = StandardScaler()

# Eğitim verisine fit et ve dönüştür

X_train[sayisal_degiskenler] = scaler.fit_transform(
    X_train[sayisal_degiskenler]
)

# Test verisini yalnızca dönüştür

X_test[sayisal_degiskenler] = scaler.transform(
    X_test[sayisal_degiskenler]
)
```

```python
# Ölçeklendirme sonrası sayısal değişkenlerin özeti

X_train[sayisal_degiskenler].describe()
```
